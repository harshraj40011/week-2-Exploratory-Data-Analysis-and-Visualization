```python
product_analysis = (
    df.groupby(["Product ID", "Product Name"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
    .reset_index()
)

product_analysis.head()

high_sales_threshold = product_analysis["Sales"].quantile(0.75)

high_sales_products = product_analysis[
    product_analysis["Sales"] >= high_sales_threshold
]

low_profit_threshold = product_analysis["Profit"].quantile(0.25)

high_sales_low_profit = high_sales_products[
    high_sales_products["Profit"] <= low_profit_threshold
].sort_values("Sales", ascending=False)

high_sales_low_profit.head(10)
```

```python
plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=product_analysis,
    x="Sales",
    y="Profit"
)

plt.axvline(
    high_sales_threshold,
    linestyle="--",
    label="High Sales Threshold"
)

plt.axhline(
    low_profit_threshold,
    linestyle="--",
    label="Low Profit Threshold"
)

plt.title("Products with High Sales but Low Profit")
plt.xlabel("Total Sales")
plt.ylabel("Total Profit")
plt.legend()

plt.show()
```
